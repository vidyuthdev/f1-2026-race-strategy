```python
import fastf1
import pandas as pd

fastf1.Cache.enable_cache('f1_cache')

high_speed_2025 = ['Silverstone', 'Spa', 'Monza', 'Suzuka', 'Jeddah',
                   'Albert Park', 'Las Vegas', 'Red Bull Ring']
all_laps = []

for track in high_speed_2025:
    print(f"Loading {track}...")
    session = fastf1.get_session(2025, track, 'R')
    session.load(telemetry=False, weather=True)
    laps = session.laps[['Driver','LapNumber','Compound','TyreLife','Stint',
                         'LapTime','PitInTime','PitOutTime','TrackStatus','IsAccurate']].copy()
    laps['LapTimeS'] = laps['LapTime'].dt.total_seconds()
    laps['Track'] = track
    all_laps.append(laps)

df = pd.concat(all_laps, ignore_index=True)
print(f"\nTotal laps pulled: {len(df)}")
print(df.groupby('Track').size())
df.head()
```

```python
def get_pit_stop_times(df):
    """
    For each pit stop, compute time lost = pit lap time minus the driver's
    normal (median green-lap) pace in that race. Returns a dataframe of stops.
    """
    stops = []
    for (track, driver), g in df.groupby(['Track', 'Driver']):
        g = g.sort_values('LapNumber')
        # driver's normal pace this race = median of non-pit, valid laps
        normal = g[(g['Pitted'] == 0) & (g['LapTimeS'].notna())]['LapTimeS'].median()
        # every pit-affected lap
        for _, lap in g[g['Pitted'] == 1].iterrows():
            if pd.notna(lap['LapTimeS']):
                loss = lap['LapTimeS'] - normal
                if 0 < loss < 60:   # sanity bounds
                    stops.append({
                        'Track': track, 'Driver': driver,
                        'Lap': lap['LapNumber'], 'PitLapTime': lap['LapTimeS'],
                        'NormalPace': normal, 'TimeLost': loss
                    })
    return pd.DataFrame(stops)

pit_times = get_pit_stop_times(df)
print(f"Found {len(pit_times)} pit stops")
print(f"Average time lost per pit lap: {pit_times['TimeLost'].mean():.2f}s")
pit_times.head(15)
```

```python
df_dry = df[~df['Compound'].isin(['INTERMEDIATE', 'WET'])].copy()
print(f"After dropping wet: {len(df)} -> {len(df_dry)}")
```

```python
def clean_laps(df):
    d = df.copy()
    d = d[d['LapTimeS'].notna()]                          # must have a time
    d = d[d['IsAccurate'] == True]                        # FastF1 accuracy flag
    d = d[d['Pitted'] == 0]                               # no pit laps
    d = d[d['LapNumber'] > 1]                             # no lap 1
    def green(ts): return all(c not in str(ts) for c in ['4','5','6','7'])
    d = d[d['TrackStatus'].apply(green)]                  # no safety-car laps
    def keep_fast(g): return g[g['LapTimeS'] <= 1.07 * g['LapTimeS'].median()]
    d = d.groupby('Track', group_keys=False).apply(keep_fast)  # no outliers
    return d

clean = clean_laps(df_dry)
print(f"After full cleaning: {len(df_dry)} -> {len(clean)}")
clean.head()
```

```python
import numpy as np

def fit_degradation_per_track(df):
    results = {c: [] for c in ['SOFT','MEDIUM','HARD']}
    for track, tdf in df.groupby('Track'):
        for comp in ['SOFT','MEDIUM','HARD']:
            sub = tdf[tdf['Compound']==comp].copy()
            # drop any rows with missing/inf values in the fit columns
            sub = sub[['LapTimeS','TyreLife','LapNumber']].replace([np.inf,-np.inf], np.nan).dropna()
            if len(sub) < 50:
                continue
            # skip if tyre age never varies (can't fit a slope)
            if sub['TyreLife'].nunique() < 2:
                continue
            X = np.column_stack([np.ones(len(sub)),
                                 sub['TyreLife'].to_numpy(float),
                                 sub['LapNumber'].to_numpy(float)])
            y = sub['LapTimeS'].to_numpy(float)
            try:
                b0, d, fuel = np.linalg.lstsq(X, y, rcond=None)[0]
            except np.linalg.LinAlgError:
                print(f"  [skip] {track} {comp}: fit didn't converge")
                continue
            results[comp].append({'track':track,'base':b0,'d':d,'fuel':fuel,'n':len(sub)})

    params = {}
    print(f"\n{'Comp':6s} {'avg d':>8s} {'avg fuel':>9s} {'tracks':>7s}")
    for comp in ['SOFT','MEDIUM','HARD']:
        fits = results[comp]
        if not fits:
            print(f"{comp}: no track had enough clean laps"); continue
        d_avg    = np.mean([f['d'] for f in fits])
        fuel_avg = np.mean([f['fuel'] for f in fits])
        base_avg = np.mean([f['base'] for f in fits])
        params[comp] = {'base':base_avg,'d':d_avg,'fuel':fuel_avg,
                        'n_tracks':len(fits),'per_track':fits}
        print(f"{comp:6s} {d_avg:+8.4f} {fuel_avg:+9.4f} {len(fits):>7d}")
    return params

params = fit_degradation_per_track(clean)

# show which tracks contributed, per compound
for comp in params:
    print(f"\n{comp} fitted on:")
    for f in params[comp]['per_track']:
        print(f"  {f['track']:15s} d={f['d']:+.4f} fuel={f['fuel']:+.4f} (n={f['n']})")
```

```python
def measure_pit_cost(df):
    green_losses, sc_losses = [], []
    for (track, driver), g in df.groupby(['Track','Driver']):
        g = g.sort_values('LapNumber')
        # driver's normal pace this race
        normal = g[(g['Pitted']==0) & g['LapTimeS'].notna()]['LapTimeS'].median()
        if pd.isna(normal): continue
        for _, lap in g[g['Pitted']==1].iterrows():
            if pd.isna(lap['LapTimeS']): continue
            loss = lap['LapTimeS'] - normal
            if not (0 < loss < 60): continue          # sanity bounds
            under_sc = any(c in str(lap['TrackStatus']) for c in ['4','6','7'])
            (sc_losses if under_sc else green_losses).append(loss)

    import numpy as np
    result = {
        'green_pit_loss': float(np.median(green_losses)) if green_losses else None,
        'sc_pit_loss': float(np.median(sc_losses)) if sc_losses else None,
        'n_green': len(green_losses), 'n_sc': len(sc_losses)
    }
    print(f"Green pit loss (median): {result['green_pit_loss']:.1f}s  (n={result['n_green']})")
    if result['sc_pit_loss']:
        print(f"SC pit loss (median):    {result['sc_pit_loss']:.1f}s  (n={result['n_sc']})")
    else:
        print(f"SC pit loss: no safety-car pit laps found (n=0)")
    return result

pit_cost = measure_pit_cost(df)   # use full df (includes pit laps), not clean
```

```python
# Diagnose: separate in-laps from out-laps, and check the distribution
import numpy as np

in_losses, out_losses = [], []
for (track, driver), g in df.groupby(['Track','Driver']):
    g = g.sort_values('LapNumber')
    normal = g[(g['Pitted']==0) & g['LapTimeS'].notna()]['LapTimeS'].median()
    if pd.isna(normal): continue
    for _, lap in g.iterrows():
        if pd.isna(lap['LapTimeS']): continue
        loss = lap['LapTimeS'] - normal
        if not (0 < loss < 60): continue
        if pd.notna(lap['PitInTime']):  in_losses.append(loss)
        if pd.notna(lap['PitOutTime']): out_losses.append(loss)

print(f"IN-laps (entering pit):  median {np.median(in_losses):.1f}s  (n={len(in_losses)})")
print(f"OUT-laps (exiting pit):  median {np.median(out_losses):.1f}s  (n={len(out_losses)})")
print(f"Implied TOTAL stop cost: {np.median(in_losses)+np.median(out_losses):.1f}s")
```

```python
import requests, time

OPENF1 = "https://api.openf1.org/v1"

sessions = requests.get(f"{OPENF1}/sessions",
            params={"year": 2025, "session_name": "Race"}, timeout=30).json()

total, with_sc, with_vsc = 0, 0, 0
for s in sessions:
    key = s['session_key']; country = s.get('country_name','?')
    rc = requests.get(f"{OPENF1}/race_control",
                      params={"session_key": key}, timeout=30).json()
    if not isinstance(rc, list): 
        continue
    msgs = [str(m.get('message','')).upper() for m in rc if isinstance(m, dict)]
    has_sc  = any('SAFETY CAR' in m and 'VIRTUAL' not in m for m in msgs)
    has_vsc = any('VIRTUAL SAFETY CAR' in m for m in msgs)
    total += 1
    with_sc += has_sc
    with_vsc += has_vsc
    print(f"  {country:18s}  SC={'Y' if has_sc else '-'}  VSC={'Y' if has_vsc else '-'}")
    time.sleep(0.3)

print(f"\nTotal races: {total}")
print(f"Races with SC:  {with_sc}  ({with_sc/total:.0%})")
print(f"Races with VSC: {with_vsc}  ({with_vsc/total:.0%})")

p_sc_per_race = with_sc/total
```

```python
import json
import numpy as np

# Convert per-race probability to per-lap.
# If p_race = chance of >=1 SC in a race of N laps, then:
#   p_race = 1 - (1 - p_lap)^N   ->   p_lap = 1 - (1 - p_race)^(1/N)
AVG_LAPS = 57   # typical high-speed race length; refine per circuit later

p_sc_race = 13/24     # 0.54
p_sc_lap = 1 - (1 - p_sc_race) ** (1/AVG_LAPS)

print(f"P(SC per race): {p_sc_race:.2f}")
print(f"P(SC per lap):  {p_sc_lap:.4f}   (~{p_sc_lap*100:.1f}% each lap)")

# Save to parameters.json
safety_car = {
    'p_sc_per_race': p_sc_race,
    'p_sc_per_lap': p_sc_lap,
    'p_vsc_per_race': 9/24,
    'avg_sc_duration_laps': 4,     # typical; cite Aguad (SC ~3-4 laps)
    'n_races': 24,
    'source': 'measured 2025 (24 races); cross-checks Aguad & Thraves ~47%'
}
params_file = json.load(open('parameters.json'))
params_file['safety_car'] = safety_car
json.dump(params_file, open('parameters.json','w'), indent=2)
print("\nSaved:", json.dumps(safety_car, indent=2))
```
